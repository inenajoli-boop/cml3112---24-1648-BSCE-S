## q2: 4 marks
Write percent_error(measured, expected) returning the signed percent error.

```python
def percent_error(measured, expected):
    """Signed percent error: 100 * (measured - expected) / expected."""
    return 100 * (measured - expected) / expected
```

## q3: 3 marks
Write slab_mass(area_m2, thickness_m) returning the mass of the concrete slab in kg, taking concrete as 2400 kg/m3.

```python
def slab_mass(area_m2, thickness_m):
    """Mass of a concrete slab in kg (area m2 x thickness m x 2400 kg/m3)."""
    CONCRETE_DENSITY = 2400  # kg/m3
    return area_m2 * thickness_m * CONCRETE_DENSITY
```

## q4: 5 marks
Write efficiency(out_w, in_w) returning the percentage, and returning None when in_w is zero or negative.

```python
def efficiency(out_w, in_w):
    """Efficiency in % = 100 * output power / input power. None if in_w <= 0."""
    if in_w <= 0:
        return None
    return 100 * out_w / in_w
```
