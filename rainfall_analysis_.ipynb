```python
import pandas as pd
import matplotlib.pyplot as plt

# make data
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
rain = [80, 60, 120, 200, 180, 90, 60, 80, 150, 200, 180, 100]
df = pd.DataFrame({'Month': months, 'Rainfall': rain})

# show table
print(df)

# show graph
plt.plot(df['Month'], df['Rainfall'])
plt.title('Monthly Rainfall in Kampala')
plt.show()
```
